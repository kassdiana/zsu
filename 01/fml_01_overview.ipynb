💡By default, sample() returns a different random sample each time it is called

```python
df.sample(5, random_state=42)
```

💡 Setting random_state makes the random selection reproducible — the same rows will be selected each time.

.head() и .tail(): Выводят первые или последние строки. По умолчанию — 5 строк, но можно передать любое число в скобки (например, df.head(10)).

.sample(): Выдает случайные строки. Это полезно, чтобы посмотреть данные из разных частей таблицы, а не только отсортированные с начала.
Без random_state (например, просто df.sample(5)): Каждый раз, когда вы запускаете эту строчку, компьютер будет выдавать вам 5 разных строк.

С random_state=42: Компьютер зафиксирует алгоритм выбора. Если вы запустите этот код сейчас, завтра или через год на другом компьютере — вы получите одни и те же 5 строк.

Зачем это нужно?
В анализе данных и машинном обучении критически важно получать одинаковые результаты при повторных запусках одного и того же кода. Если вы делите данные на обучающую и тестовую выборки, вы хотите, чтобы при перезапуске программы (или когда ваш код запустит коллега) данные разделились точно так же, иначе результаты экспериментов будет невозможно сравнить.

### .describe() - basic statistics on numerical columns

```python
df.describe()
```

💡 By default, describe() summarizes only numerical columns. You can use <code>include="all"</code> to include all columns, including string columns.

```python
df.describe(include="all")
```

Команда df.describe() выводит базовую описательную статистику для всех числовых колонок в датафрейме. Это один из главных способов быстро посмотреть, как распределены данные.

В результате вы получите таблицу, где колонки — это числовые признаки из данных, а строки — статистические метрики:

count — количество непустых (не NaN) значений. Помогает сразу увидеть, в каких колонках не хватает данных.

mean — среднее арифметическое значение.

std (standard deviation) — стандартное отклонение. Показывает, насколько сильно значения «разбросаны» вокруг среднего. Чем больше std, тем сильнее разброс.

min — минимальное значение в колонке.

25% (первый квартиль) — значение, ниже которого находится 25% всех данных.

50% (медиана) — значение, которое делит данные пополам. Половина значений меньше медианы, половина — больше. Часто это более показательная метрика, чем mean, так как она не искажается экстремальными выбросами.

75% (третий квартиль) — значение, ниже которого находится 75% данных.

max — максимальное значение.

Зачем это нужно?
Эта команда помогает за пару секунд найти аномалии и выбросы перед началом анализа. Например:

Если в колонке «Возраст клиента» min равен -5 или max равен 150 — сразу понятно, что в данных есть ошибки.

Если count в одной колонке 1000, а в другой 850 — значит, во второй есть 150 пропущенных значений.

Важный нюанс: По умолчанию describe() игнорирует текстовые (категориальные) колонки и анализирует только числа. Если нужна статистика по текстовым колонкам (количество уникальных значений, самое частое значение), нужно написать df.describe(include='object') или df.describe(include='all').

- <code>loc</code>: selects data by **labels**

## <code>loc</code>   
### data.loc [row_labels, column_labels]

Это два главных инструмента для срезов (slicing) данных. Разница между ними принципиальна:

loc ищет по именам (меткам). Если у тебя стандартный индекс (0, 1, 2...), то loc[5] выдаст строку с названием 5. При срезах (например, 0:5) loc включает последнее значение (выдаст строки от 0 до 5 включительно).

iloc ищет по позициям (как индексы в массивах C/C++). Срез 0:5 не включает последнее значение (выдаст строки на позициях 0, 1, 2, 3, 4). Сюда можно передавать отрицательные индексы (например, -1 для последней строки или колонки).

### Quick CheatSheet

| Action                  | `loc` – By Label                        | `iloc` – By Position    |
| ----------------------- | --------------------------------------- | ----------------------- |
| Select a single row     | `data.loc[2]`                           | `data.iloc[2]`          |
| Select a single column  | `data.loc[:, "Temperature"]`            | `data.iloc[:, 3]`       |
| Select multiple rows    | `data.loc[[1, 3, 5]]`                   | `data.iloc[[1, 3, 5]]`  |
| Select multiple columns | `data.loc[:, ["Date", "Temperature"]]`  | `data.iloc[:, [0, 3]]`  |
| Select a range of rows  | `data.loc[:5]`                          | `data.iloc[:5]`         |
| Select rows and columns | `data.loc[:5, ["Date", "Temperature"]]` | `data.iloc[:5, [0, 3]]` |
| Select from the end     | —                                       | `data.iloc[-5:, -2:]`   |

### Matplotlib

```python
import matplotlib.pyplot as plt 
```

- #### Line plot

🧩 Let's visualize the average number of rented bikes for each hour of the day.

```python
plt.figure(figsize=(10, 5))
#plt.plot(hourly_demand.index, hourly_demand.values, marker="o")

plt.xlabel("Hour")
plt.ylabel("Average rented bike count")
plt.title("Average Bike Demand During the Day")

plt.show()
```

By default, Matplotlib connects the observations with a line.

We can plot multiple lines in the same figure. Each plot() call adds another line to the current plot.

```python
hourly_mean = df.groupby("Hour")["Rented_bike_count"].mean()
hourly_max = df.groupby("Hour")["Rented_bike_count"].max()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(hourly_mean.index, hourly_mean.values, label="Mean",marker="o")

plt.plot(hourly_max.index, hourly_max.values,label="Maximum", marker="^")

plt.xlabel("Hour")
plt.ylabel("Rented bike count")
plt.title("Bike Demand During the Day")
plt.legend()

plt.show()
```
